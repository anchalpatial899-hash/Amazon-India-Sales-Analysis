```python
# Which product categories generate the most revenue?

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.plot(monthly_sales.index.astype(str), monthly_sales.values, marker='o')

plt.title('Monthly Sales Trend - 2025')
plt.xlabel('Month')
plt.ylabel('Total Sales (INR)')

plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
category_sales = df.groupby('Product_Category')['Total_Sales_INR'].sum()

category_sales
```

```python
# Which product categories generate the most sales?
plt.figure(figsize=(8, 5))

plt.bar(category_sales.index, category_sales.values)

plt.title('Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales (INR)')

plt.xticks(rotation=45)
plt.grid(axis='y')

plt.show()
```

```python
state_sales = df.groupby('State')['Total_Sales_INR'].sum()

state_sales
```

```python
#Which Indian states generate the most sales, and which generate the least?
plt.figure(figsize=(10, 8))

plt.barh(state_sales_sorted.index, state_sales_sorted.values)

plt.title('Sales by State')
plt.xlabel('Total Sales (INR)')
plt.ylabel('State')

plt.grid(axis='x')
plt.show()
```

```python
# Which individual products generate the most and least sales?
product_sales = df.groupby('Product_Name')['Total_Sales_INR'].sum()

product_sales_sorted = product_sales.sort_values(ascending=False)

product_sales_sorted
```

```python
delivery_counts = df['Delivery_Status'].value_counts()

delivery_counts
```

```python
#Are returns concentrated in particular states? 
returned_by_state = (
    df[df['Delivery_Status'] == 'Returned']
    .groupby('State')
    .size()
    .sort_values(ascending=False)
)

returned_by_state
```

```python
return_rate_by_state = (
    df.groupby('State')
      .apply(lambda x: (x['Delivery_Status'] == 'Returned').mean() * 100)
      .sort_values(ascending=False)
)

return_rate_by_state
```

```python
# Which product categories have the highest return rates?
return_rate_by_category = (
    df.groupby('Product_Category')
      .apply(lambda x: (x['Delivery_Status'] == 'Returned').mean() * 100)
      .sort_values(ascending=False)
)

return_rate_by_category
```

```python
payment_counts = df['Payment_Method'].value_counts()

payment_counts
```

```python
# Does the payment method have any relationship with returned orders? 
return_rate_by_payment = (
    df.groupby('Payment_Method')
      .apply(lambda x: (x['Delivery_Status'] == 'Returned').mean() * 100)
      .sort_values(ascending=False)
)

return_rate_by_payment
```

```python
# How are customers rating their purchases?
rating_counts = df['Review_Rating'].value_counts().sort_index()

rating_counts
```

```python
# How many customers are repeat customers versus one-time customers?
customer_orders = df.groupby('Customer_ID')['Order_ID'].nunique()

customer_orders.value_counts().sort_index()
```

```python
customer_sales = (
    df.groupby('Customer_ID')['Total_Sales_INR']
      .sum()
      .sort_values(ascending=False)
)

customer_sales.head(10)
```

```python
customer_summary = (
    df.groupby('Customer_ID')
      .agg(
          Orders=('Order_ID', 'nunique'),
          Total_Sales=('Total_Sales_INR', 'sum'),
          Average_Order_Value=('Total_Sales_INR', 'mean')
      )
      .sort_values('Total_Sales', ascending=False)
)

customer_summary.head(10)
```

```python
# Which product categories generate the highest average revenue per order?
category_aov = (
    df.groupby('Product_Category')['Total_Sales_INR']
      .mean()
      .sort_values(ascending=False)
)

category_aov
```

```python
# Quantity Analysis
quantity_sales = (
    df.groupby('Quantity')['Total_Sales_INR']
      .agg(['count', 'sum', 'mean'])
)

quantity_sales
```

```python
# Average Order Value by Quantity:plt.figure(figsize=(8, 5))

plt.bar(quantity_sales.index, quantity_sales['mean'])

plt.title('Average Order Value by Quantity')
plt.xlabel('Quantity per Order')
plt.ylabel('Average Order Value (INR)')

plt.grid(axis='y')
plt.show()
```

```python
# 5. Final Visualizations
### Business Question
# What proportion of orders are delivered, pending, or returned?

delivery_counts = df['Delivery_Status'].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(delivery_counts.index, delivery_counts.values)

plt.title('Order Distribution by Delivery Status')
plt.xlabel('Delivery Status')
plt.ylabel('Number of Orders')

plt.grid(axis='y')
plt.show()
```

```python
## 5.2 Return Rate by State

### Business Question

# Which states have the highest and lowest return rates?
return_rate_by_state = (
    df.groupby('State')
      .apply(lambda x: (x['Delivery_Status'] == 'Returned').mean() * 100)
      .sort_values()
)

plt.figure(figsize=(10, 8))

plt.barh(
    return_rate_by_state.index,
    return_rate_by_state.values
)

plt.title('Return Rate by State')
plt.xlabel('Return Rate (%)')
plt.ylabel('State')

plt.grid(axis='x')
plt.show()
```

```python
## 5.3 Return Rate by Product Category

### Business Question

# Are returns concentrated in particular product categories?

return_rate_by_category = (
    df.groupby('Product_Category')
      .apply(lambda x: (x['Delivery_Status'] == 'Returned').mean() * 100)
      .sort_values()
)

plt.figure(figsize=(8, 5))

plt.bar(
    return_rate_by_category.index,
    return_rate_by_category.values
)

plt.title('Return Rate by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Return Rate (%)')

plt.xticks(rotation=45)
plt.grid(axis='y')

plt.show()
```

```python
## 5.4 Review Rating Distribution

### Business Question

# How are customers rating their purchases?

rating_counts = df['Review_Rating'].value_counts().sort_index()

plt.figure(figsize=(8, 5))

plt.bar(
    rating_counts.index,
    rating_counts.values
)

plt.title('Review Rating Distribution')
plt.xlabel('Review Rating')
plt.ylabel('Number of Reviews')

plt.xticks([1, 2, 3, 4, 5])
plt.grid(axis='y')

plt.show()
```

```python
## 5.5 Average Order Value by Quantity.

# How does the quantity purchased in an order affect its average order value?

plt.figure(figsize=(8, 5))

plt.bar(quantity_sales.index, quantity_sales['mean'])

plt.title('Average Order Value by Quantity')
plt.xlabel('Quantity per Order')
plt.ylabel('Average Order Value (INR)')

plt.grid(axis='y')
plt.show()
```

```python
## 6.1 Overall Sales Performance

- Total sales were approximately ₹111.82 crore across 15,000 orders.
- Monthly sales remained relatively stable throughout 2025, ranging from approximately ₹8.50 crore to ₹9.76 crore.
- August recorded the highest monthly sales, while February recorded the lowest.
- The data does not show a strong continuous upward or downward sales trend.

## 6.2 Product Category Performance

- Sales were fairly evenly distributed across the five product categories.
- Beauty generated the highest category sales at approximately ₹22.75 crore.
- Home & Kitchen generated the lowest at approximately ₹21.67 crore.
- The relatively small difference suggests that no single category dominates the dataset.

 ## 6.3 Regional Performance

- Sikkim recorded the highest sales at approximately ₹4.31 crore.
- Gujarat recorded the lowest sales at approximately ₹3.62 crore.
- Sales were distributed relatively evenly across the 28 states.
- Regional performance varies, but there is no single state that overwhelmingly dominates sales.

## 6.4 Product-Level Performance

- Lipstick was the highest-selling product by total sales at approximately ₹4.82 crore.
- Vacuum Cleaner recorded the lowest product sales at approximately ₹4.12 crore.
- The relatively narrow gap indicates that sales are distributed across a broad product portfolio
```

```python
# 7. Operational & Customer Insights


 ## 7.1 Delivery Performance

- 5,075 orders were delivered, 5,044 were pending, and 4,881 were returned.
- Only about one-third of orders were marked as delivered in the dataset.
- The high proportion of pending and returned orders represents a potential operational area for further investigation.

 ## 7.2 Return Analysis

- Return rates varied across states, ranging from approximately 28.44% to 36.35%.
- Assam recorded the highest return rate, while Maharashtra recorded the lowest.
- Return rates across product categories were relatively similar, ranging from approximately 31.27% to 33.90%.
- Therefore, state-level differences appear more noticeable than category-level differences in this dataset.
```

```python
  
 # 8. Business Recommendations

Based on the analysis, the following areas could be investigated:

1. Investigate states with relatively high return rates to identify possible operational or customer-experience issues.
2. Analyze pending orders to understand whether delays are concentrated in particular states, products, or periods.
3. Investigate the reasons behind product returns rather than relying only on return counts.
4. Focus on improving customer experience because the average review rating is approximately 3.04/5.
5. Analyze high-value and repeat customers to identify opportunities for customer retention and personalized offers.
6. Monitor lower-performing products and regions for potential improvement opportunities.
```
